```python
import anthropic

claude = anthropic.Anthropic(api_key="paste your api key here")

def generate_performance_review(manager_notes, employee_name, role, review_period):
    response = claude.messages.create(
        model="claude-haiku-4-5-20251001",
        max_tokens=2000,
        messages=[{
            "role": "user",
            "content": f"""You are an expert HR coach helping managers write fair, structured, and evidence-based performance reviews in Singapore.

Manager's raw notes about the employee:
{manager_notes}

Employee name: {employee_name}
Role: {role}
Review period: {review_period}

Generate a structured performance review in this exact format:

## PERFORMANCE REVIEW
**Employee:** {employee_name}
**Role:** {role}
**Period:** {review_period}

## OVERALL RATING: [Choose one: Exceptional / Exceeds Expectations / Meets Expectations / Needs Improvement / Unsatisfactory]

## KEY ACHIEVEMENTS
(3-4 specific achievements with evidence from the manager's notes)

## AREAS FOR DEVELOPMENT
(2-3 specific development areas with coaching suggestions)

## COMPETENCY RATINGS
- Delivery & Results: [1-5] — [one line justification]
- Communication: [1-5] — [one line justification]
- Teamwork & Collaboration: [1-5] — [one line justification]
- Initiative & Innovation: [1-5] — [one line justification]
- Leadership (if applicable): [1-5] — [one line justification]

## COACHING FRAMEWORK
(3 specific, actionable coaching recommendations for the next review period with suggested timelines)

## SUGGESTED GOALS FOR NEXT PERIOD
(3 SMART goals based on development areas)

## MANAGER SUMMARY
(2-3 sentences summarising overall performance and potential)"""
        }]
    )
    print(response.content[0].text)

# Test with sample manager notes
manager_notes = """
Sarah has been really good this year. She delivered the Q3 campaign on time even though the brief kept changing.
She's great with clients but sometimes goes quiet in team meetings. I think she lacks confidence.
She helped onboard two new junior staff which was above her job scope.
There were a few times she missed deadlines on internal reports but always got client work done.
She wants to move into a senior role next year. I think she could get there but needs to work on speaking up more
and being more proactive with internal stakeholders. Her technical skills are strong.
"""

generate_performance_review(
    manager_notes=manager_notes,
    employee_name="Sarah Tan",
    role="Marketing Executive",
    review_period="January to June 2026"
)
```
